```python
arr = np.array([12, 25, 8, 40, 15, 60, 5, 35])
print(arr[arr>20])
print(arr[arr<15])
print(arr[arr%2==0])
print(arr[(arr>=10) & (arr<=40)])
```

```python
arr = np.array([5, 12, 18, 25, 32, 40, 45, 55, 60])
print(arr[(arr<15) | (arr>50)])
print(arr[(arr>20) & (arr<45)])
print(arr[arr!=25])
print(arr[(arr%2!=0) & (arr>15)])
```
