#### aggrgation function

```python
import numpy as np
a=np.array([10,20,30,40])
print(np.sum(a)) #to get sum of all elements
print(a.sum()) #to get sum of all elements
print(np.mean(a)) #to get mean of all elements
print(a.mean()) #to get mean of all elements
print(np.median(a)) # it will sort the array and then find the middle value to get median of all elements
print(a.median()) #it will sort the array and then find the middle value to get median of all elements
print(np.std(a)) #to get standard deviation of all elements
print(a.std()) #to get standard deviation of all elements
print(np.var(a)) #to get variance of all elements
print(np.min(a)) #to get minimum value of all elements
print(a.min()) #to get minimum value of all elements 
print(np.max(a)) #to get maximum value of all elements
print(a.max()) #to get maximum value of all elements        
print(np.argmin(a)) #to get index of minimum value of all elements
print(a.argmin()) #to get index of minimum value of all elements
print(np.argmax(a)) #to get index of maximum value of all elements
print(a.argmax()) #to get index of maximum value of all elements
```

### sorting
* 1d sorting
* 2d sorting

```python
import numpy as np
a=np.array([2,0,7,8,6,5,4,3,1,9])
a.sort() #it will sort the array in ascending order
print(a)
print(a[::-1]) #it will sort the array in descending order
```

```python
arr=np.array([[4,1],[2,3]])
arr.sort() #it will sort the rows of the array in ascending order
print(arr)
arr1=arr.flatten() #it will flatten the array 2d array into 1D array
print(arr1)
arr1.sort() #it will sort the flattened array in ascending order
print(arr1)
arr=arr1.reshape(2,2) #it will reshape the flattened array 1d array  into 2d array 2 rows and 2 columns
print(arr)
```

### filtering


```python
x=np.array([1,2,10,20])
print(x>=10) #to get boolean values
print(x[x>=10]) #to get values greater than or equal to 10
print((x>=10) & (x<20)) #to get boolean values between 10 and 20
print(x[(x>=10) & (x<20)]) #to get values between 10 and 20
print((x>=10) | (x>20)) #to get boolean values greater than or equal to 10 or greater than 20
print(x[(x>=10) | (x>20)]) #to get values greater than or equal to 10 or greater than 20
```

```python
x=np.array([1,2,10,20])
y=np.where(x>10) #it will return the indices of the elements that satisfy the condition
print(y)
y=np.where(x>10, 1, 0) #it will replace the values greater than 10 with 1 and the values less than or equal to 10 with 0
print(y)
#in where we can give three parameters first is the condition, second is the value to be replaced if the condition is true and third is the value to be replaced if the condition is false
```

```python
marks=np.array([50,60,100,20,30,40,80,90])
print(np.where(marks<50,0,marks))
```

### unique number
