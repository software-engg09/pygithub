### filtering data

```python
import pandas as pd
data={
    'id':[1,2,3,4,5],
    'name':['shazeb','ammar','zayed','awais','ishaq'],
    'marks':[38,73,91,52,81],
    'age':[22,21,20,9,50],
    'department':['it','hr','hr','it','it']}
df=pd.DataFrame(data) # it will create a dataframe with the given data 
df
```

```python
print(df[df['marks'] > 50]) # it will return the rows of the dataframe where marks are greater than 50
df['marks']> 50 # it will return the boolean values of the dataframe where marks are greater than 50
```

```python
print(df[df['marks'] > 50][['name','age']]) # it will return the rows of the dataframe where marks are greater than 50 and it will return the name and age columns of the dataframe
```

```python
print(df.loc[df['marks'] > 50, ['name','age']]) # it will return the rows of the dataframe where marks are greater than 50 and it will return the name and age columns of the dataframe
```

```python
df.loc[df['name'].str.startswith('a'),['id','name']] # it will return the rows of the dataframe where name starts with 'a'  
```

```python
df.loc[df['name'].str.len() > 5, ['id','name']] # it will return the rows of the dataframe where name length is greater than 5
```

```python
df.loc[(df['marks'] > 50) & (df['department'] == 'it')] # it will return the rows of the dataframe where marks are greater than 50 and it will return the name and department columns of the dataframe
```

```python
df.loc[(df['marks'] > 50) & (df['department'].isin(['it','hr']))] # it will return the rows of the dataframe where marks are greater than 50 and department is either 'it' or 'hr' and it will return the name and department columns of the dataframe
```

```python
df.loc[(df['marks'] > 50) & ~(df['department'].isin(['it']))]  # it will return the rows of the dataframe where marks are greater than 50 and department is not 'it' and it will return the name and department columns of the dataframe
```

```python
df.loc[df['marks'].between(50,73),['id','name']]# it will return the rows of the dataframe where marks are between 50 and 73 inclusive including 50 and 73
```

### apply function

```python
def grade(marks):
    if marks >80:
        return 'A'
    elif marks >60:
        return 'B'
    elif marks >40:
        return 'C'
    else:
        return 'fail'
print(grade(90))
```

```python
df['grade']=df['marks'].apply(grade) #to apply the funtion inthe pandas
df
```

```python
df['bonus']=df['marks'].apply(lambda x:x+5)
df
```

```python
df.drop(columns='bonus',inplace=True)
df
```

### join in python(merge)

```python
students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','adnan']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,30]})
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='inner'
) 
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='right'
) 
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='left'
) 
# to get sepecific column use df['column_name]
df[['name','marks']]
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='outer'
) 
# the outer is the both left join and right join
```

```python
s = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','adnan']})
m = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,30]})
```

```python
# to merge the two df with having diff column names we can use left_on and right_on
d=pd.merge(
    s,
    m,
    left_on='student_id',
    right_on='id',
    how='left'
) 
d[['name','student_id','marks']]
```
